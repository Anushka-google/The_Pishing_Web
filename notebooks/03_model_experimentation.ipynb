# Phase 8: Model Experimentation & Comparative Analysis
## Phishing Detection & Risk Intelligence Platform

This notebook trains and benchmarks 4 candidate classifiers on the engineered URL feature matrix:
1. **Logistic Regression** (Linear baseline)
2. **Decision Tree** (Interpretable tree)
3. **Random Forest** (Bagging ensemble)
4. **XGBoost** (Gradient boosted trees)

> *Select the final model based on experimental evidence, not because XGBoost is popular.*

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from src.training.model_experimentation import ModelExperimenter

experimenter = ModelExperimenter(features_csv='../data/processed/features.csv')
results, champion, champion_path = experimenter.run_experiments()

df_results = pd.DataFrame(results)
display(df_results[['model', 'precision', 'recall', 'f1', 'roc_auc', 'pr_auc', 'accuracy', 'train_time_sec', 'latency_ms_per_url']])
```

### 1. Model Performance Comparison Across Metrics

```python
models = [r['model'] for r in results]
f1_scores = [r['f1'] * 100 for r in results]
accuracies = [r['accuracy'] * 100 for r in results]

x = np.arange(len(models))
width = 0.35

fig, ax = plt.subplots(figsize=(9, 5))
rects1 = ax.bar(x - width/2, accuracies, width, label='Accuracy %', color='#2ecc71')
rects2 = ax.bar(x + width/2, f1_scores, width, label='F1-Score %', color='#3498db')

ax.set_ylabel('Score (%)')
ax.set_title('Model Performance: Accuracy & F1-Score Comparison')
ax.set_xticks(x)
ax.set_xticklabels(models, rotation=10)
ax.set_ylim([90, 102])
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### 2. Training Time & Inference Latency

```python
for r in results:
    print(f"{r['model']:22s} | Train Time: {r['train_time_sec']:.4f}s | Latency: {r['latency_ms_per_url']:.5f} ms/URL")
print(f"\nSelected Champion: {champion['model']}")
```
