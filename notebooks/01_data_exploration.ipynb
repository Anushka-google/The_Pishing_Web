# Phase 3: Exploratory Data Analysis (EDA)
## Phishing Detection & Risk Intelligence Platform

This notebook conducts an in-depth exploratory data analysis of the ingested URL dataset:
- Class distribution
- URL and domain length distributions
- Structural and character anomalies
- Data quality and duplicate audit
- Domain overlap and data leakage risks
- Suspicious lexical keyword distribution

```python
import pandas as pd
import numpy as np
import json
from src.preprocessing.eda import DataExplorer

# Initialize EDA engine on processed dataset
explorer = DataExplorer(data_path='../data/processed/urls.csv')
report = explorer.run_full_eda()
print("Total records analyzed:", report['class_distribution']['total_records'])
print("Phishing percentage:", report['class_distribution']['phishing_pct'], "%")
print("Legitimate percentage:", report['class_distribution']['legitimate_pct'], "%")
```

### 1. Length & Structural Characteristics

```python
length_df = pd.DataFrame(report['length_distributions']).T
display(length_df)

structural_df = pd.DataFrame(report['structural_distributions']).T
display(structural_df)
```

### 2. Data Leakage & Domain Overlap Risks

```python
leakage = report['data_leakage_and_domains']
print("Unique Phishing Domains:", leakage['unique_phishing_domains'])
print("Unique Legitimate Domains:", leakage['unique_legitimate_domains'])
print("Cross-Class Overlapping Domains:", leakage['overlapping_domain_count'])
print("Top Repeated Domains:", leakage['top_repeated_domains'])
print("\nLeakage Warning:", leakage['leakage_risk_assessment'])
```

### 3. Suspicious Lexical Signals

```python
keyword_df = pd.DataFrame(report['keyword_signals']).T
display(keyword_df.sort_values(by='phishing_rate_pct', ascending=False))
```
