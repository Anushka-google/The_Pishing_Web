# Phase 4 (Research Approaches): Empirical Benchmark of Detection Paradigms
## Phishing Detection & Risk Intelligence Platform

This notebook demonstrates and compares the four major phishing detection paradigms on the dataset:
1. **Blacklist-based Detection** (Hash/URL lookup from threat intelligence feeds)
2. **Heuristic-based Detection** (Static regex and syntactic pattern matching)
3. **Machine Learning-based Detection** (Static URL structural and lexical features)
4. **Hybrid Detection Engine** (Multi-layered consensus combining Blacklist fast-path + Heuristic guardrails + ML score)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import json
from src.evaluation.benchmark_approaches import ApproachesBenchmark

# Run the benchmark suite
benchmark = ApproachesBenchmark(dataset_path='../data/processed/urls.csv')
results = benchmark.run_benchmark()

df_results = pd.DataFrame(results)
display(df_results[['paradigm', 'accuracy', 'precision', 'recall', 'zero_day_recall', 'false_negative_rate', 'avg_latency_ms']])
```

### 1. Comparison of Detection Recall: Overall vs. Zero-Day Phishing

```python
paradigms = [r['paradigm'].split('.')[1].strip().split('(')[0] for r in results]
recall_all = [r['recall'] * 100 for r in results]
recall_zero_day = [r['zero_day_recall'] * 100 for r in results]

x = np.arange(len(paradigms))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 5))
rects1 = ax.bar(x - width/2, recall_all, width, label='Overall Recall %', color='#3498db')
rects2 = ax.bar(x + width/2, recall_zero_day, width, label='Zero-Day Recall %', color='#e74c3c')

ax.set_ylabel('Recall Percentage (%)')
ax.set_title('Phishing Detection Rate: Overall vs. Zero-Day Threat Generalization')
ax.set_xticks(x)
ax.set_xticklabels(paradigms, rotation=15)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### 2. Latency vs. Protection Trade-Off Analysis

```python
latencies = [r['avg_latency_ms'] for r in results]
f1_scores = [r['f1_score'] * 100 for r in results]

for p, lat, f1 in zip(paradigms, latencies, f1_scores):
    print(f"{p:30s} | Avg Latency: {lat:6.4f} ms | F1-Score: {f1:5.1f}%")
```
